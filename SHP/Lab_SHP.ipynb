```python
conn = psycopg2.connect(
    host="localhost",
    database="Data_Warehouse",
    user="postgres",
    password="123456"
)

query = """
SELECT
    t.mes,
    SUM(f.monto) AS ventas
FROM fact_ventas f
INNER JOIN dim_tiempo t
ON f.id_tiempo = t.id_tiempo
GROUP BY t.mes
ORDER BY t.mes
"""

df = pd.read_sql_query(query, conn)

plt.plot(
    df["mes"],
    df["ventas"],
    marker="o"
)

plt.title("Ventas por mes")
plt.show()

 
```

```python
conn = psycopg2.connect(
    host="localhost",
    database="Data_Warehouse",
    user="postgres",
    password="123456"
)

query = """
SELECT
    p.producto,
    SUM(f.cantidad) AS total_vendido
FROM fact_ventas f
INNER JOIN dim_producto p
ON f.id_producto = p.id_producto
GROUP BY p.producto
ORDER BY total_vendido DESC
"""

df = pd.read_sql_query(query, conn)

plt.figure(figsize=(8,5))
plt.bar(df["producto"], df["total_vendido"])
plt.title("Productos Más Vendidos")
plt.xlabel("Producto")
plt.ylabel("Cantidad Vendida")
plt.show()

conn.close()
```

```python
import psycopg2
import pandas as pd
import matplotlib.pyplot as plt

conn = psycopg2.connect(
    host="localhost",
    database="Data_Warehouse",
    user="postgres",
    password="123456"
)

query = """
SELECT
    p.producto,
    SUM(f.cantidad) AS total_vendido
FROM fact_ventas f
INNER JOIN dim_producto p
ON f.id_producto = p.id_producto
GROUP BY p.producto
ORDER BY total_vendido DESC
"""

df = pd.read_sql_query(query, conn)

plt.figure(figsize=(8,5))
plt.bar(df["producto"], df["total_vendido"])
plt.title("Productos Más Vendidos")
plt.xlabel("Producto")
plt.ylabel("Cantidad Vendida")
plt.show()

conn.close()
```

Conclusiones
La implementación de este Data Warehouse utilizando PostgreSQL, Python y Jupyter Notebook ha permitido comprender en profundidad los conceptos teóricos y prácticos del almacenamiento analítico de datos. A lo largo de este proyecto se observaron los siguientes aprendizajes clave:
•	Un Data Warehouse no es simplemente una base de datos más grande; su diseño orientado al análisis (esquema estrella, desnormalización) lo diferencia radicalmente de una BD operacional.
•	El proceso ETL representa la etapa más crítica del DW. La calidad de los análisis depende directamente de la limpieza y transformación correcta de los datos en esta fase.
•	Python, a través de pandas y SQLAlchemy, ofrece herramientas poderosas y sencillas para automatizar la carga de datos desde múltiples fuentes hacia el DW.
•	Jupyter Notebook democratiza el acceso al análisis de datos, permitiendo combinar código, visualizaciones y narrativa en un documento reproducible y compartible.
•	Las visualizaciones generadas (barras, líneas, pastel, heatmap) son fundamentales para comunicar hallazgos a tomadores de decisiones sin conocimientos técnicos.
•	PostgreSQL demostró ser una solución sólida y completamente funcional para implementar un DW a escala académica, con soporte completo para SQL analítico y operaciones de agregación.

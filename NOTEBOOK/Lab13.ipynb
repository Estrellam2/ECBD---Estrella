```python
#Importar librerias 
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report, ConfusionMatrixDisplay)
```

```python
# Cargamos el dataset Iris
from sklearn.datasets import load_iris
iris = load_iris(as_frame = True)
```

```python
print(iris.DESCR)
```

```python
dfiris = iris.frame
```

```python
# Renombrar la columna target
dfiris.rename(columns = {" target": "especie" }, inplace = True)
```

```python
dfiris["especie"] = dfiris["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})
```

```python
dfiris.head()
```

```python
dfiris.tail()
```

## Identificar variables predictoras y variables objetivo

```python
# variable predictora
x = dfiris.drop(columns = ["especie"])
```

```python
print("Variables predictoras")
print(x.columns)
```

```python
#Variable objetivo
y= dfiris["especie"]
```

```python
print("Variables objetivo")
print(y.name)
```

## Dividir los datos en prueba y entrenamiento

80% para entrenamiento

20% para prueba

```python
x_train, x_test, y_train, y_test= train_test_split(
    x, y, test_size= 0.20, random_state= 42, stratify = y
)
```

```python
print("x_train", x_train.shape)
print("x_test", x_test.shape)
print("y_train", y_train.shape)
print("y_test", y_test.shape)
```

## Escalado de variables

```python
scaler = StandardScaler()
```

```python
#escalamos las variables
x_train_escalado = scaler.fit_transform(x_train)
x_test_escalado = scaler.transform(x_test)
```

```python
print("Sin escalar")
print(x_train.head())
print("Escalado")
print(x_train_escalado[:5])
```

```python
modelo= KNeighborsClassifier(n_neighbors = 3)
```

```python
modelo.fit(
    x_train_escalado, y_train
)
```

## Realizamos predicciones con las variables de prueba

```python
prediccion = modelo.predict(x_test_escalado)
```

```python
print(prediccion[:10])
```

```python
# Comparación entre valores rales valores predecidos
comparacion = pd.DataFrame({
    "especie real":y_test.values,
    "esecie predicha":prediccion
})
```

```python
comparacion.head(30)
```

# Evaluación del modelo

```python
accuracy = accuracy_score(y_test, prediccion)
print(f"accuracy/exactitud: {accuracy: .4f}")
```

```python
cm = confusion_matrix(y_test,prediccion)
```

```python
cm_df = pd.DataFrame(
    cm,
    index = modelo.classes,
    columns= modelo.classes_
)
cm_df
```
